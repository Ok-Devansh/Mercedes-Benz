```python
import pandas as pd 
df = pd.read_csv("Sales Data.csv")
data = pd.DataFrame(df)
df.info()
df.describe()
```

```python
df.columns = df.columns.str.lower()
print(df.columns)
```

```python
df.isnull().sum()
```

```python
import numpy as np
df = df.replace(['null', 'NULL', 'Null', np.nan],' ').fillna(0)
```

```python
df.isnull().sum()
```

```python
df = df.replace({'P003x':'P003',
                'D069x' :'D069'}
                ).fillna('NA') 
```

```python
df['product_id'] = df['product_id'].str.strip()
df['dealer_id'] = df['dealer_id'].str.strip()
```

```python
df['customer_id'] = df['customer_id'].str.strip()
```

```python
print(df.head(35))
```

```python
columns = ["sales","discount_%","quantity","unit_price"]

for col in columns:
    df[col] = pd.to_numeric(df[col], errors= "coerce").astype("Int64")
```

```python
df.info()
```

```python
import pandas as pd
df2 = pd.read_csv("Dim_Customer_2026_csv.csv")
data = pd.DataFrame(df2)
df2.isnull().sum()
```

```python
df2.columns = df2.columns.str.lower()
print(df2.columns)
```

```python
df2['customer_id'] = df2['customer_id'].str.strip()
```

```python
import numpy as np
df2 = df2.dropna()
```

```python
df2.isnull().sum()
```

```python
print(df2.head(35))
```

```python
df2 = df2.replace({'MALE':'Male',
                   'male':'Male',
                   'FEMALE':'Female',
                   'female':'Female'}
                 ).fillna('NA')
```

```python
cols = ['customer_id','customer_name','gender','country']

for col in cols:
    df2[col] = df2[col].astype(str).str.strip().str.capitalize()
```

```python
print(df2.head(35))
```

```python
print(df2.head(35))
```

```python
df2['age']= pd.to_numeric(df2['age'],errors= 'coerce').astype('Int64')
```

```python
df2.info()
```

```python
print(df2.head(35))
```

```python
labels = ['Young', 'Young-adult', 'Senior','Senior-aged']
df2['age_group'] = pd.qcut(df2['age'], q=4, labels = labels)
```

```python
print(df2)
```

```python
df4.columns = df4.columns.str.lower()
print(df4.columns)
```

```python
data1 = pd.merge(df,df1,on= 'dealer_id')
data2 = pd.merge(df,df2,on= 'customer_id')
data3 = pd.merge(df,df3,on= 'product_id')
data4 = pd.merge(df,df4, on = 'date_id')
```

```python
gender_wise_sale = data2.groupby('gender')['sales'].sum()
```

```python
print(gender_wise_sale)
```

```python
country_wise_sale = data1.groupby('country')['sales'].sum()
print(country_wise_sale)
```

```python
agegroup_wise_sale = data2.groupby('age_group')['sales'].sum()
print(agegroup_wise_sale)
```

```python
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]

Month_wise_sales = (
    data4.groupby('month_name')['sales'].sum().reindex(month_order))

print(Month_wise_sales)
```

```python
quarter_wise_sales = data4.groupby('quarter')['sales'].sum()
print(quarter_wise_sales)
```

```python
category_wise_sales = data3.groupby('category')['sales'].sum()
print(category_wise_sales)
```

```python
category_wise_profit = (data3.groupby('category')['profit'].sum() / data1['profit'].sum() * 100).round(2)
print(category_wise_profit)
```

```python
fuel_wise_profit = (data3.groupby('fuel_type')['profit'].sum() / data1['profit'].sum() * 100).round(2)
print(fuel_wise_profit)
```

```python
model_wise_sales = data3.groupby('model_name')['sales'].sum()
print(model_wise_sales)
```
